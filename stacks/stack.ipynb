Stacks

```python
class stack():
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)
        print("Pushed item is", item)

    def pop(self):
        if self.items:
            item = self.items.pop()
            print("Popped item is", item)
            return item
        else:
            print("Stack is empty")
            return None
    
    def peek(self):
        if self.items:
            top  = len(self.items)  
            print("Top element is", top)
            return top

    def is_empty(self):
        if self.items:
            return False
        else:
            return True

    def size(self):
        return len(self.items)

    def display(self):
        for i in range(len(self.items)):
            print(self.items[i])



st = stack()

st.push(1)
st.push(2)
st.push(3)
st.push(4)
st.push(5)
st.push(6)


print(st.pop())
print(st.pop())
print(st.peek())
print(st.is_empty())
print(st.size())    

st.display()

            
```

Stack Using Linked List

```python
# USING LINKED LIST

class stack():
    def __init__(self,val):
        self.data = val
        self.next = None
    
    def push(self,val):
        nd = stack(val)
        if self.next is None:
            self.next = nd
        else:
            c = self
            while c.next is not None:
                c = c.next
            c.next = nd
            
        return self.next
    
    def pop(self):
        if self.next is None:
            return None
        else:
            c = self
            while c.next is not None:
                c = c.next

            self.next = c.next
            return c.data
        
    def display(self):
        c = self
        while c.next is not None:
            print('---',c.data,end=" ")
            c = c.next
        print("NONE")   

    def isEmpty(self):
        if self.next is None:
            return True
        else:
            return False
    
    def peek(self):
        if self.next is None:
            return None
        else:
            return self.next.data
        

st = stack(5)
st.push(10)
st.push(20)
st.push(30)
st.push(40)

print(st.pop())
print(st.pop())
print(st.peek())
print(st.isEmpty())

st.display()

    

      
```

Balanced Parentheses
Given a string s, check whether the parentheses in s are balanced. If the parentheses are balanced, return true. Otherwise, return false.

```python
import queue
def balance(s):

    stack = queue.LifoQueue()



    if len(s) == 0:
        print("Stack is empty")
        return True
    else:
        for i in s:
            if i in '({[':
                print('iiii',i)
                stack.put(i)
            elif i in ')}]':
                print('iiii',i)
                if stack.empty():
                    print("Stack is empty---")
                    return False
                if i == ')':
                    if stack.get() != '(':
                        print("Parenthesis mismatch")
                        return False
                elif i == ']':
                    if stack.get() != '[':
                        print("parentesis mismatch1")
                        return False
                elif i == '}':
                    if stack.get() != '{':
                        print("parentesis mismatch2")
                        return False
            print('--',i)
        return True
        

    



s = '((a+b))*((c+d)))' 
print(balance(s))

```
